# Dutch definition audit
One row per word in release-9500. This reproduces mechanical checks, not a native-speaker language review.
Sources: pinned release CSV, definitions overlay, and explicit editorial drafts. All files are local and checksummed by the build script.

```python
from pathlib import Path
import csv, json, sys
root = Path.cwd()
while not (root / "scripts/word_definitions.py").exists():
    if root.parent == root:
        raise RuntimeError("Run inside the Woordenboom repository")
    root = root.parent
sys.path.insert(0, str(root / "scripts"))
import word_definitions as definitions
original = definitions.release.read_csv(definitions.SOURCE)
rows = definitions.release.read_csv(definitions.DEST / "words.csv")
editorial = json.loads((definitions.DEST / "editorial.json").read_text())
definitions.validate(rows, original)
assert rows == definitions.build_rows(original, editorial)
report = definitions.quality(rows, original, editorial)
print(json.dumps(report, indent=2))
```
